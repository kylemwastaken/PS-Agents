```python

import heapq

# 1. Define the directed highway graph
graph = {
    "Nairobi": {
        "Naivasha": 89.5,
        "Narok": 141.5,
        "Thika": 44
    },

    "Naivasha": {
        "Nakuru": 71
    },

    "Nakuru": {
        "Eldoret": 156,
        "Kericho": 107
    },

    "Eldoret": {
        "Kitale": 71
    },

    "Kitale": {
        "Kakamega": 110
    },

    "Kakamega": {
        "Kisumu": 50
    },

    "Narok": {
        "Bomet": 81.5
    },

    "Bomet": {
        "Kericho": 74.25
    },

    "Kericho": {
        "Sondu": 45,
        "Kisii": 87.2
    },

    "Kisii": {
        "Sondu": 61.3
    },

    "Sondu": {
        "Kisumu": 52.6
    },

    "Thika": {
        "Nyeri": 110
    },

    "Nyeri": {
        "Nyahururu": 104
    },

    "Nyahururu": {
        "Nakuru": 64
    }
}


# 2. Uniform Cost Search algorithm
def uniform_cost_search(graph, start, goal):

    # Priority queue stores (path_cost, current_node)
    frontier = [(0, start)]

    # Track the cheapest known cost to each node
    best_cost = {start: 0}

    # Track parent nodes to reconstruct the path
    parent = {start: None}

    # Record the order in which nodes are removed
    explored = []

    while frontier:

        # Get the node with the lowest path cost
        current_cost, current_node = heapq.heappop(frontier)

        # Ignore outdated queue entries
        if current_cost > best_cost[current_node]:
            continue

        explored.append((current_node, current_cost))

        print(f"Exploring: {current_node}, "
              f"Path Cost: {current_cost} km")

        # Check whether the goal has been reached
        if current_node == goal:

            # Reconstruct the optimal path
            path = []
            node = goal

            while node is not None:
                path.append(node)
                node = parent[node]

            path.reverse()

            return path, current_cost, explored

        # Explore neighbouring locations
        for neighbour, edge_cost in graph[current_node].items():

            new_cost = current_cost + edge_cost

            # Update only if a cheaper path is found
            if (neighbour not in best_cost or
                    new_cost < best_cost[neighbour]):

                best_cost[neighbour] = new_cost
                parent[neighbour] = current_node

                heapq.heappush(
                    frontier,
                    (new_cost, neighbour)
                )

    return None, float("inf"), explored


# 3. Define the initial and goal states
start = "Nairobi"
goal = "Kisumu"

print("UNIFORM COST SEARCH")
print("-------------------")

# 4. Execute the search
optimal_path, total_cost, explored_nodes = (
    uniform_cost_search(graph, start, goal)
)

# 5. Display the results
print("\nSEARCH RESULTS")
print("-------------------")

if optimal_path:

    print("Optimal Path:")
    print(" -> ".join(optimal_path))

    print(f"\nTotal Path Cost: {total_cost} km")

    print("\nNodes Explored:")
    for node, cost in explored_nodes:
        print(f"{node}: {cost} km")

else:
    print("No path found between the two locations.")
```
